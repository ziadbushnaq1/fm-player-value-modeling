This project is a application of a player valuation model for Football Manager 24, a soccer management simulation video game where you act as a manager of a soccer team and run all aspects of the club. There is plenty of data available to build a model to find undervalued players. I set the game to run 10 years in the future to January 15th, 2035 to allow for real-life players to retire and generated players to populate the game. I collected all players who had total transfer fees collected over their career > 50,000 GBP, and collected all their attributes, stats, and general information.

To test out my player valuation model, I wanted to be at a club in Germany, as they have lax player registration rules for foreign players. Clubs in the top division were not interested in taking a chance on an inexperienced manager, so I looked at roles in the second division.


 Magdeb






The data hub showed that as a team we were underperforming the league average in goals per game, non-penalty expected goals created per 90 minutes, shots per game, and shots on target percentage. This indicates that we are lacking in quality in attacking positions.





Looking at the depth charts, to address the underperformance we should target players for the left wing and striker positions. Although we are doing well defensively, the depth chart shows a strong need for a new left back. 

After initial squad assessment and looking over team stats, I decided to target LW, ST, LB. to address loan gap and reinforce squad holes. I ran the player value model function to find the best value players for these positions in the database that fit some financial limits.

From these center back options, Petrazzuolo and Limoeiro had contracts coming up and were willing to join at a reasonable wage.


We ended up signing 5 free agents to come in during the summer to refresh the squad. Ted Curd came in as prospective signings and our own players showed concerns that we were lacking quality goalkeeper depth. Correia and Robinson came in as versatile attacking options that could both cover left wing and striker positions. Limeoria and Petrazzuolo came in at center backs to raise the level in that position and to free up wage space from other high earners we could potentially move in. We did not address the gap at left back this window, as I didn't find a player that fit the tactics at a reasonable salary. To account for lack of depth there, I converted a left-footed midfielder that was low in the pecking order in his natural position.

In the first six months as the manager of Magdeburg, I took the team from 12th place to 4th. Unfortunately, we narrowly missed out qualification for the promotion playoff which required 3rd place.
